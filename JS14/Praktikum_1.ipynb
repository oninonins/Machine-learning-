Import Library

```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

Pra Pengolahan Data

 Pra Pengolahan Data Training

```python
train_datagen = ImageDataGenerator(rescale = 1./255,
	                                   shear_range = 0.2,
	                                   zoom_range = 0.2,
	                                   horizontal_flip = True)
training_set = train_datagen.flow_from_directory('/content/drive/MyDrive/dataset ml /dataset/training_set',
	                                                 target_size = (64, 64),
	                                                 batch_size = 32,
	                                                 class_mode = 'binary')
```

Pra Pengolahan Data Testing

```python
test_datagen = ImageDataGenerator(rescale = 1./255)
test_set = test_datagen.flow_from_directory('/content/drive/MyDrive/dataset ml /dataset/test_set',
	                                            target_size = (64, 64),
	                                            batch_size = 32,
	                                            class_mode = 'binary')
```

Pembuatan Model CNN

Inisiasi Model CNN

```python
cnn = tf.keras.models.Sequential()
```

Pembuatan Layer Konvolusi 1

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=[64, 64, 3]))
```

Pembuatan Layer Pooling 1

```python
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

Pembuatan Layer Konvolusi 2 dan Pooling 2

```python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
```

Flattening

```python
cnn.add(tf.keras.layers.Flatten())
```

Fully Connected Layer 1 (Input)

```python
cnn.add(tf.keras.layers.Dense(units=128, activation='relu'))
```

Fully Connected Layer 2 (Output)

```python
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid'))
```

Compile Model CNN

```python
cnn.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

 Fit CNN

```python
cnn.fit(x = training_set, validation_data = test_set, epochs = 25)
```

Prediksi dengan 1 Citra

```python
import numpy as np
from keras.preprocessing import image
test_image = image.load_img('/content/drive/MyDrive/dataset ml /dataset/single_prediction/cat_or_dog_1.jpg', target_size = (64, 64))
test_image = image.img_to_array(test_image)
test_image = np.expand_dims(test_image, axis = 0)
result = cnn.predict(test_image)
training_set.class_indices
if result[0][0] == 1:
  prediction = 'dog'
else:
  prediction = 'cat'
```

1. **Model mencapai akurasi 100% pada data latih dan data uji.** Setelah Epoch 1, akurasi latih dan validasi sama-sama 1.0000 dan bertahan sampai Epoch 25.

2. **Validasi loss sangat rendah, lebih rendah daripada training loss.** Pada Epoch 1, `val_loss` sudah **4.1266e-08** dan terus turun hingga **6.0152e-15** pada Epoch 25. Pola ini tidak wajar: training loss biasanya lebih rendah daripada validation loss karena augmentasi dan dropout hanya aktif saat training.

3. **Indikasi kuat data leakage atau duplikasi gambar.** Dengan hanya **231 gambar latih**, wajar jika model belum sempurna pada Epoch 1 (akurasi latih 97.84%), tetapi tidak wajar jika validation loss langsung mendekati nol. Kemungkinan besar data uji memuat gambar yang sama atau hampir identik dengan data latih, misalnya frame berurutan dari video yang sama. Akurasi 100% karena itu **tidak dapat dianggap sebagai bukti generalisasi**.

4. **Ukuran data uji lebih besar daripada data latih.** Data latih 231 gambar, data uji 1.000 gambar. Perbandingan ini tidak lazim dan memperkuat dugaan bahwa pembagian data bukan pembagian independen.

5. **Training sangat cepat setelah epoch pertama.** Epoch 1 memerlukan sekitar 270 detik karena inisialisasi, sedangkan epoch berikutnya sekitar 7-12 detik. Total 25 epoch selesai dalam beberapa menit.

6. **Tidak ada informasi prediksi akhir pada output.** Cell prediksi menjalankan `cnn.predict(test_image)` dan menyimpan hasil ke variabel `prediction`, tetapi nilainya tidak dicetak. Output hanya menampilkan progres prediksi `1/1`, sehingga label cat atau dog untuk citra tunggal tidak dapat diverifikasi dari notebook.

7. **Overfitting tidak dapat dinilai dari kurva ini.** Kedua kurva sudah menyentuh 1.0000 sejak awal, sehingga tidak ada informasi tentang jarak train dan validasi. Evaluasi yang valid memerlukan data uji yang benar-benar terpisah.

### Kesimpulan

1. Model CNN berhasil mencapai akurasi **100%** pada data latih dan data uji yang dipakai notebook.
2. Hasil tersebut **belum dapat disimpulkan sebagai performa generalisasi yang baik** karena validation loss yang sangat kecil mengindikasikan kebocoran data atau duplikasi antara data latih dan data uji.
3. Untuk evaluasi yang benar, data perlu dipisahkan ulang secara independen, misalnya 80% latih dan 20% uji dari total gambar yang tidak beririsan, lalu data uji hanya dipakai sekali di akhir.
